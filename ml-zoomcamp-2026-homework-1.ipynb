## Getting the data

Reads the local file if present, otherwise downloads it from the course repository.

```python
import os

fname = 'car_fuel_efficiency_2026.csv'
url = 'https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/' + fname

df = pd.read_csv(fname if os.path.exists(fname) else url)
df.head()
```

## Q2. Records count

```python
len(df)
```

## Q3. Fuel types

```python
print(df['fuel_type'].value_counts())
df['fuel_type'].nunique()
```

## Q4. Missing values

```python
missing = df.isnull().sum()
print(missing)
print()
print('Columns with missing values:', (missing > 0).sum())
```

## Q5. Max fuel efficiency of cars from Asia

```python
df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].max()
```

## Q6. Median value of horsepower

```python
median_before = df['horsepower'].median()
most_frequent = df['horsepower'].mode()[0]
median_after = df['horsepower'].fillna(most_frequent).median()

print('Median before:', median_before)
print('Most frequent value:', most_frequent)
print('Median after fillna:', median_after)
```

## Q7. Sum of weights

```python
X = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7).to_numpy()

XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y

print('w =', w)
print('Sum of w =', w.sum())
```
